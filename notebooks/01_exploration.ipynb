```python
df = df.drop(columns=["EmployeeCount", "Over18", "StandardHours"])
df.shape
```

```python
import matplotlib.pyplot as plt

df.groupby("OverTime")["Attrition"].value_counts(normalize=True).unstack().plot(kind="bar", color=["#1B3A5C", "#E8735A"])
plt.title("Attrition Rate by OverTime Status")
plt.ylabel("Proportion")
plt.savefig("../outputs/charts/overtime_attrition.png", bbox_inches="tight")
plt.show()
```

```python
dept_attrition = df.groupby("Department")["Attrition"].value_counts(normalize=True).unstack()
dept_attrition.plot(kind="bar", figsize=(8,5), color=["#1B3A5C", "#E8735A"])
plt.title("Attrition Rate by Department", fontsize=13, fontweight="bold")
plt.ylabel("Proportion")
plt.xlabel("Department")
plt.xticks(rotation=15)
plt.legend(title="Attrition")
plt.tight_layout()
plt.savefig("../outputs/charts/department_attrition.png", bbox_inches="tight")
plt.show()
```

```python
income_by_attrition = df.groupby("Attrition")["MonthlyIncome"].mean()
income_by_attrition.plot(kind="bar", figsize=(6,5), color=["#1B3A5C", "#E8735A"])
plt.title("Average Monthly Income: Stayed vs Left", fontsize=13, fontweight="bold")
plt.ylabel("Average Monthly Income")
plt.xlabel("Attrition")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../outputs/charts/income_attrition.png", bbox_inches="tight")
plt.show()
```

```python
satisfaction_by_attrition = df.groupby("Attrition")["JobSatisfaction"].mean()
satisfaction_by_attrition.plot(kind="bar", figsize=(6,5), color=["#1B3A5C", "#E8735A"])
plt.title("Average Job Satisfaction: Stayed vs Left", fontsize=13, fontweight="bold")
plt.ylabel("Average Job Satisfaction (1-4)")
plt.xlabel("Attrition")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("../outputs/charts/satisfaction_attrition.png", bbox_inches="tight")
plt.show()
```
